# 00 — SNP / data prefetch

Thin wrapper over `scripts/snp.py`. For every basket in `scripts/baskets.py`
it calls `SNP(...).cached_fetch(name=basket)`, which downloads from yfinance
on first run and writes `data/<basket>.parquet` + `data/<basket>_sample.csv`
to Drive (Colab) or the local repo. No plots — EDA lives in `01_eda` for the
combined 16-equity universe and in `00_visuals` for the motivational figures.

Re-running this is safe and cheap: the cache is invalidated only if the
parquet's tickers or date range no longer matches the requested window.

> **Runtime:** ~30 s on Colab cold (4 baskets × yfinance fetch); ~3 s on cache.

### IMPORTS

```python
# === Bootstrap (Colab + local) ===
import os, urllib.request as _u
exec((open('../scripts/bootstrap.py') if os.path.exists('../scripts/bootstrap.py') else _u.urlopen('https://raw.githubusercontent.com/egil10/fys5419/main/project2/code/scripts/bootstrap.py')).read())

# === Project imports ===
from scripts.snp     import SNP
from scripts.baskets import get, BASKETS

START_DATE = "2023-01-01"
END_DATE   = "2025-12-31"
print(f'baskets to prefetch: {BASKETS}')
```

### DOWNLOAD

```python
# cached_fetch downloads on first run and writes:
#   data/<basket>.parquet       (full price history)
#   data/<basket>_sample.csv    (first 20 rows for quick inspection)
# Re-runs hit the parquet cache unless tickers or date window changed.
for name in BASKETS:
    SNP(get(name), START_DATE, END_DATE).cached_fetch(name=name)
```
