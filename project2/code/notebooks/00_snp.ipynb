# 00 — SNP / data fetch

Single responsibility: download all ticker baskets defined in `scripts/baskets.py`
to `data/` (Drive on Colab, local repo otherwise). Produces one per-basket
overview plot in `plots/snp/` as a sanity check.

> **Runtime:** ~30 s on Colab CPU (~5 s if parquet caches already exist).

### IMPORTS

```python
# === Bootstrap (Colab + local) ===
import os, urllib.request as _u
exec((open('../scripts/bootstrap.py') if os.path.exists('../scripts/bootstrap.py') else _u.urlopen('https://raw.githubusercontent.com/egil10/fys5419/main/project2/code/scripts/bootstrap.py')).read())

# === Project imports ===
from scripts.colab   import out_dir
from scripts.snp     import SNP
from scripts.baskets import get, BASKETS

PLOTS_DIR = out_dir('plots', 'snp')
print(f'Plots will be saved to: {PLOTS_DIR}')

START_DATE = "2023-01-01"
END_DATE   = "2025-12-31"
```

### DOWNLOAD

```python
# Fetch + cache each basket, then write a per-basket price panel to plots/snp/.
for name in BASKETS:
    snp = SNP(get(name), START_DATE, END_DATE).cached_fetch(name=name)
    snp.summary()
    snp.plot(save=True, name=name, individual=False,
             figsize=(12, 6), save_dir=PLOTS_DIR)
```
