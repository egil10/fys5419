# Classical baselines

Brute force, greedy, simulated annealing, and Markowitz-rounded — all on the
same cardinality-constrained QUBO defined in `portfolio.py`. Brute force gives
us a ground-truth label everywhere `n` is small enough.

### SETUP

```python
import sys
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import pandas as pd

from scripts.data      import load_returns
from scripts.baskets   import config
from scripts.portfolio import Portfolio
from scripts.classical import run_all
```

### PROBLEM

```python
BASKET = 'mag7'
K, LAM, AP = 2, 2.0, 0.5

tickers, start, end = config(BASKET)
r  = load_returns(tickers, start, end, cache_name=BASKET)
pf = Portfolio(r.mu, r.Sigma, lam=LAM, A=AP, K=K, tickers=list(r.tickers))
print(pf)
```

### RUN ALL BASELINES

```python
results = run_all(pf)

rows = [{
    'solver':   r.name,
    'bitstring': r.bitstring,
    'tickers':  '+'.join(r.tickers(pf)),
    'cost':     r.cost,
    'feasible': r.feasible,
    'n_evals':  r.n_evals,
    'runtime_s': r.runtime,
} for r in results.values()]
pd.DataFrame(rows).sort_values('cost')
```
