# Final comparison

Headline figure for the report: classical baselines vs QAOA on the same
cost function, across baskets. Pulls from `results/*.json` produced by
`depth.ipynb` and `scaling.ipynb` — no heavy computation in this notebook,
just plotting.

### SETUP

```python
import sys, json
from pathlib import Path
sys.path.append(str(Path.cwd().parent))

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scripts.plotting import apply_style, PALETTE, title, fig_path
apply_style()

RESULTS_DIR = Path.cwd().parent / 'results'
```

### DEPTH SWEEP — approximation ratio and P(optimal) vs p

```python
BASKET = 'mag7'
sweep  = pd.DataFrame(json.loads((RESULTS_DIR / f'depth_sweep_{BASKET}.json').read_text()))

fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))
axes[0].plot(sweep.p, sweep.ratio, 'o-', color=PALETTE['blue'], lw=2, ms=8)
axes[0].axhline(1.0, color=PALETTE['red'], ls='--', lw=1.2, label='optimum')
axes[0].set_xticks(sweep.p); axes[0].set_xlabel('depth p')
axes[0].set_ylabel('approximation ratio')
title(axes[0], 'QAOA approximation ratio', f'basket = {BASKET}')
axes[0].legend()

axes[1].plot(sweep.p, sweep.p_optimal, 'o-', color=PALETTE['red'], lw=2, ms=8,
             label='P(optimum)')
axes[1].plot(sweep.p, sweep.p_feasible, 'o-', color=PALETTE['blue_muted'], lw=2,
             ms=8, label='P(feasible)')
axes[1].set_xticks(sweep.p); axes[1].set_xlabel('depth p')
axes[1].set_ylabel('probability')
title(axes[1], 'Measurement probabilities', 'higher = more concentrated')
axes[1].legend()

plt.tight_layout()
fig.savefig(fig_path('compare', f'depth_{BASKET}'), bbox_inches='tight')
plt.show()
```

### SIZE SCALING — cost ratio vs n, classical vs QAOA

```python
scale = pd.DataFrame(json.loads((RESULTS_DIR / 'size_scaling.json').read_text()))

fig, ax = plt.subplots(figsize=(10, 5))
for solver, sub in scale.groupby('solver'):
    ax.plot(sub.n, sub.ratio, 'o-', lw=2, ms=8, label=solver)
ax.axhline(1.0, color=PALETTE['charcoal'], ls=':', lw=1)
ax.set_xlabel('problem size n'); ax.set_ylabel('cost / brute-force cost')
title(ax, 'Cost ratio vs problem size',
      'Lower is better; 1.0 = brute-force optimum')
ax.legend()
plt.tight_layout()
fig.savefig(fig_path('compare', 'scaling'), bbox_inches='tight')
plt.show()
```
